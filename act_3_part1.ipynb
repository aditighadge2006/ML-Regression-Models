```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
data = {
    'Area': [500, 700, 900, 1100, 1300, 1500, 1700, 1900, 2100, 2300],
    'Price': [25, 32, 40, 48, 55, 65, 72, 82, 90, 100]
}

df = pd.DataFrame(data)

print(df)
```

```python
print(df.head())
print(df.info())
print(df.isnull().sum())
```

```python
X = df['Area'].values
Y = df['Price'].values

print("X =", X)
print("Y =", Y)
```

```python
plt.scatter(X, Y)

plt.xlabel("House Area (sq.ft)")
plt.ylabel("House Price (₹ lakh)")
plt.title("House Area vs House Price")

plt.show()
```

```python
X = X.reshape((-1,1))
Y = Y.reshape((-1,1))

print(X)
print(Y)
```

```python
from sklearn.linear_model import LinearRegression

Model = LinearRegression()

print(Model)
```

```python
Model.fit(X, Y)
```

```python
print("Intercept =", Model.intercept_)
```

```python
print("Slope =", Model.coef_)
```

```python
test = np.array([[1800]])

prediction = Model.predict(test)

print("Predicted Price for 1800 sq.ft house =", prediction[0][0], "lakh")
```

```python
Y_pred = Model.predict(X)

print("Actual Prices:")
print(Y)

print("Predicted Prices:")
print(Y_pred)
```

```python
plt.scatter(X, Y)

plt.plot(X, Y_pred)

plt.xlabel("House Area (sq.ft)")
plt.ylabel("House Price (₹ lakh)")
plt.title("Linear Regression - House Price Prediction")

plt.show()
```

```python
from sklearn.metrics import mean_absolute_error
from sklearn.metrics import mean_squared_error
from sklearn.metrics import r2_score
```

```python
MAE = mean_absolute_error(Y, Y_pred)

print("Mean Absolute Error =", MAE)
```

```python
MSE = mean_squared_error(Y, Y_pred)

print("Mean Squared Error =", MSE)
```

```python
RMSE = np.sqrt(MSE)

print("Root Mean Squared Error =", RMSE)
```

```python
R2 = r2_score(Y, Y_pred)

print("R2 Score =", R2)
```

```python
print("R2 Score =", Model.score(X, Y))
```
